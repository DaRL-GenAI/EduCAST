# EduHarness — Stage 1: Main Agent

先跑 `01_colab_setup.ipynb`。本 notebook 会自动从 Colab Secret **`openai`** 读取 API Key。

产出：`MyDrive/eduharness/runs/<request_id>/blueprint/`


```python
from __future__ import annotations
from pathlib import Path

from eduharness.colab_session import ensure_api_key, load_session
from eduharness.config import Config
from eduharness.pipeline import load_request, stage1

session = load_session()
ensure_api_key()  # 每个 notebook 独立内核，必须重新加载 API Key

REPO, RUN_DIR = Path(session["REPO"]), Path(session["RUN_DIR"])
REQUEST_PATH = Path(session.get("REQUEST", REPO / "examples" / "sample_request.json"))
NOTES = REPO / "examples" / "binary_notes.txt"

request = load_request(REQUEST_PATH)
if not request.source_text and NOTES.is_file() and "binary" in REQUEST_PATH.name:
    request.source_text = NOTES.read_text(encoding="utf-8")

cfg = Config.from_env(request, run_dir=RUN_DIR / request.request_id)
blueprint = stage1(cfg)
print("style theme:", blueprint.style.theme)
print("palette:", blueprint.style.palette.model_dump())
print("saved:", cfg.blueprint_path)
```
